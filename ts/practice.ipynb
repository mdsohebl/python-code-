```python
def s(st):
    l=[]
    for i in sorted(st.split("-")):
        l.append(i)
    return "-".join(l)
        
    
st="green-red-yellow-black-white"
s(st)
```

```python
def st(s):
    u=0
    l=0
    for i in s:
        if i.isupper():
            u+=1
        elif i.islower():
            l+=1
    return u,l
    
s='CampusX is an Online Mentorship Program fOr EnginEering studentS.'
u,l=st(s)
print(u)
print(l)
```

```python
def even(l):
    l1=[]
    for i in l:
        if i%2==0:
            l1.append(i)
    return l1
l=[1, 2, 3, 4, 5, 6, 7, 8, 9]
even(l)
            
```

```python
def per(num):
    s=0
    for i in range(1,num):
        if num%i==0:
            s+=i
    if s==num:
        return "its a perfect number"
    else:
        return "its not perfect number"
per(6)
```

```python
def up(*kwargs):
    d={}
    for i in kwargs:
        d.update(i)
    return d
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
up(dic1,dic2,dic3)
```

```python
def ss(s):
    d={}
    for i in s.split():
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    maxv=max(d.values())
    for i in d:
        if d[i]==maxv:
            print(i,"-->",d[i])
            break
        
s="hello how  are are are you i am fine thank you"
ss(s)
```
