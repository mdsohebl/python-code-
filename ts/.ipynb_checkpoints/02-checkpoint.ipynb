```python
class Atm:
    def __init__(self):
        self.pin=""
        self.balan=0
        self.menu()
    def menu(self):
        user_int=input("""
        How can i help you
        1. press 1 to creat a pin
        2. press 2 to change pin
        3. press 3 to check the balance
        4. press 4 to withdraw 
        5. to exite""")
        
        if user_int=="1":
            self.creat_pin()
        elif user_int=="2":
            self.change_pin()
        elif user_int=="3":
            self.check_bal()
        elif user_int=="4":
            self.withdraw()
        else:
            exit()
    def creat_pin(self):
        enter_pin=input("Enter a pin")
        self.pin=enter_pin
        
        enter_bal=eval(input("Enter balance"))
        self.balan=enter_bal
        
        print("pin created successfully")
        self.menu()
    
    def change_pin(self):
        enter_old_pin=input("Enter a old pin")
        if enter_old_pin==self.pin:
            enter_new_pin=input("Enter a new pin")
            self.pin=enter_new_pin
            print("pin change sucessfully")
            self.menu()
        else:
            print("incorrect old pin")
            self.menu()
        
    def check_bal(self):
        enter_pin=input("Enter a pin")
        if enter_pin==self.pin:
            print("Your current balance is",self.balan)
        else:
            print("please enter correct pin")

    def withdraw(self):
        enter_pin=input("Enter a pin")
        if enter_pin==self.pin:
            enter_amount=eval(input("Emter a amount"))
            if enter_amount<=self.balan:
                self.balan=self.balan-enter_amount
                print("withdraw sucessfully",self.balan)
            else:
                print("you have no more amount")
        else:
            print("Enter correct pin")
        
a=Atm()
```
