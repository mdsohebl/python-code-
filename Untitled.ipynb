```python
class Per:
    def __init__(self,fn,mn,ln):
        self.fname=fn
        self.mname=mn
        self.lname=ln
p=Per("soheb","luqman","jaggga")
```

```python
import json
def show(p):
    if isinstance(p,Per):
        return f"first {p.fname} middle {p.mname} lname{p.lname}"
with open("c.txt","w") as f:
    json.dump(p,f,default=show)
```

```python
with open("c.txt","r") as f:
    print(json.load(f))
```
