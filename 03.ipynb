```python
# example of speed 
import time

L = list(range(1000000))
T = tuple(range(1000000))

start = time.time()
for i in L:
     i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
    i*5
print('Tuple time',time.time()-start)   # tuple are faster than list
```

```python
# *args and **kwargs are special Python keywords that are used to pass the variable length 
# of arguments to a function

# *args:  (arbitrary argument)
# allows us to pass a variable number of non-keyword arguments to a function.

# **kwargs (arbitrary keyword argument)
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

# Points to remember while using *args and **kwargs
# order of the arguments matter(normal -> *args -> **kwargs)
# The words “args” and “kwargs” are only a convention, you can use any name of your choice
```

```python
# create range function
class My_Rangr:
    def __init__(self,s,e):
        self.start=s
        self.end=e
    def __iter__(self):
        return Range_Iterater(self)
    
```

```python
class Range_Iterater:
    def __init__(self,io):
        self.iter_ob=io
    def __iter__(self):
        return self
    def __next__(self):
        if self.iter_ob.start>=self.iter_ob.end:
            raise StopIteration
        current=self.iter_ob.start
        self.iter_ob.start+=1
        return current
```

```python
o=My_Rangr(1,11)
for i in o:
    print(i)
```

```python
# easy of implenantation
# range fun using genrater
def Gen_Range(s,e):
    for i in range(s,e):
        yield i
r=Gen_Range(10,20)
print(next(r))
print(next(r))
for i in r:
    print(i)
```

```python
#Chaining Generators
def fibonacci_numbers(nums):
    x, y = 0, 1
    for _ in range(nums):
        x, y = y, x+y
        yield x

def square(nums):
    for num in nums:
        yield num**2

print(sum(square(fibonacci_numbers(10))))
```
