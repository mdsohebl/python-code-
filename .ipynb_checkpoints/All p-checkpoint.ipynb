```python
# list comprehension

l1=[1,2,3]        # using + operation
l2=["a","b","c"]
print(l1+l2)

for i in l2:      # using append method
    l1.append(i)
print(l1)
    


l1=[1,2,3]        # using extend method
l2=["a","b","c"]
l1.extend(l2)
l1
```

```python
# cartesian products -> List comprehension on 2 lists together
l1 = [1,2,3,4]
l2 = [5,6,7,8]
for i in l1:
    for j in l2:
        print(i*j)
        
[i*j for i in l1 for j in l2]
```

```python
# speed test
import time

l=list(range(1000000))
t=tuple(range(1000000))

start=time.time()
for i in l:
    i*5
print("list time is",time.time()-start)

start1=time.time()
for j in t:
    j*5
print("tuple time is",time.time()-start1)
```

```python
# memory test
import sys

l=list(range(1000000))
t=tuple(range(1000000))

print("list size is ",sys.getsizeof(l))
print("tuple size is",sys.getsizeof(t))
```
