```python
# Write a program to find the euclidean distance between two coordinates.
# Take both the coordinates from the user as input
x1=eval(input("Enter a number"))
y1=eval(input("Enter a number"))
x2=eval(input("Enter a number"))
y2=eval(input("Enter a number"))

d=((x2-x1)**2+(y2-y1)**2)**0.5
print("The euclidian distance is",round(d,2))
```

```python
# Write a program to find the simple interest when the value
# of principle,rate of interest and time period is provided by the user.

p=eval(input("Enter a amount"))
t=eval(input("Enter a time period"))
r=eval(input("Enter a rate"))

r=(p*t*r)/100
print("The simple interest is ",r)
```

```python
# Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. 
# Assume all inputs are provided by the user
# an = a + (n – 1)d
f=eval(input("Enter a first terms"))
s=eval(input("Enter a second terms"))
n=eval(input("Enter a number"))
d=s-f
r=f+(n-1)*d
print(r)
```

```python
# Given 2 fractions, find the sum of those 2 fractions.
# Take the numerator and denominator values of the fractions from the user.
n1=eval(input("Enter a num1"))
d1=eval(input("Enter a den1"))
n2=eval(input("Enter a num2"))
d2=eval(input("Enter a den2"))

nr=n1*d2 + n2*d1
dr=d1*d2
print(f"{nr}/{dr}")
print(nr/dr)
```

```python
# Given the height, width and breadth of a milk tank, you have to find out 
# how many glasses of milk can be obtained? Assume all the inputs are provided by the user.

# Dimensions of the milk tank
# H = 20cm, L = 20cm, B = 20cm

# Dimensions of the glass
# h = 3cm, r = 1cm
```

```python
import math 

th=eval(input("Enter a height of tank"))
tw=eval(input("Enter a width of tank"))
tb=eval(input("Enter a breath of tank"))

gh=eval(input("Enter a glass height"))
gr=eval(input("Enter a glass radius"))

tank=th*tw*tb
glass=3.14*gr*gr*gh

# print((tank//glass))
print("The glass obtained ",math.floor(tank/glass))
```

```python
# Advantages of List Comprehension
#     More time-efficient and space-efficient than loops.
#     Require fewer lines of code.
#     Transforms iterative statement into a formula.
```

```python
l=[]
for i in range(1,11):
    l.append(i)
print(l)
```

```python
[i for i in range(1,11)]
```

```python
# scaler multiplication on vactor
l=[1,2,3]
s=-3
[s*i for i in l]
```

```python
# add square
l=[2,3,4]

# both are same
[i*i for i in l]  
[i**2 for i in l]
```

```python
# Print all numbers divisible by 5 in the range of 1 to 50
# for i in range(1,51):
#     if i%5==0:
# #         print(i)
[i for i in range(1,51) if i%5==0]
```

```python
# find languages which start with letter p
languages = ['java','python','php','c','javascript']
for i in languages:
    if i.startswith("p"):
        print(i)
[i for i in languages if i.startswith("p")]
```

```python
# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'
# using neasted if
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']
for i in my_fruits:
    if i in basket:
        if i.startswith("a"):
            print(i)

[i for i in basket if i in my_fruits if i.startswith("a")]
```

```python
[[i*j for i in range(1,4)] for j in range(1,4)]

    
```

```python
# cartesian products -> List comprehension on 2 lists together
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
```

```python
# 2 ways to traverse list 
# 1 iterwise 
l=[1,2,3,4]
for i in l:
    print(i)
```

```python
# indexwise
l=[1,2,3,4]
for i in range(len(l)):
    print(l[i])
```

```python
# The zip() function returns a zip object, which is an iterator of tuples where the first item in each
# passed iterator is paired together, and then the second item in each passed iterator are paired together.

# If the passed iterators have different lengths, the iterator with the least items decides 
# the length of the new iterator.
```

```python
# Write a program to add items of 2 lists indexwise
l1 = [1,2,3,4]
l2 = [5,6,7,8]

list(zip(l1,l2))

l=[]
for i,j in zip(l1,l2):
    l.append(i+j)
print(l)

[i+j for i,j in zip(l1,l2)]
```

```python
# Create 2 lists from a given list where
# 1st list will contain all the odd numbers from the original list and
# the 2nd one will contain all the even numbers

l= [1,2,3,4,5,6]
even=[]
odd=[]
for i in l:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print(even)
print(odd)
```
