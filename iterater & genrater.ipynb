```python
# create own range fun
class My_Range:
    def __init__(self,s,e):
        self.start=s
        self.end=e
    def __iter__(self):
        return Range_Iterater(self)
```

```python
class Range_Iterater:
    def __init__(self,io):
        self.iter_obj=io
    
    def __iter__(self):
        return self

    def __next__(self):
        if self.iter_obj.start >= self.iter_obj.end:
            raise StopIteration
        current=self.iter_obj.start
        self.iter_obj.start+=1
        return current
```

```python
x=My_Range(1,11)
```

```python
for i in x:
    print(i)
```

```python
# genrator
# ex
def genra():
    yield "first"
    yield "second"
    yield "third"
a=genra()
print(next(a))
print(next(a))

for i in genra():
    print(i)
```

```python
# ex 2
def square(n):
    for i in range(1,n+1):
        yield i*i

a=square(10)
print(next(a))
print(next(a))
print(next(a))
for i in a:
    print(i)
```

```python
#Representing Infinite Streams
def all_even():
    n=0
    while True:
        yield n
        n+=2
for i in all_even():
    print(i)
```

```python
#Chaining Generators
def fibonacci_numbers(nums):
    x, y = 0, 1
    for _ in range(nums):
        x, y = y, x+y
        yield x

def square(nums):
    for num in nums:
        yield num**2

print(sum(square(fibonacci_numbers(10))))
```
