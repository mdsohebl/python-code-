```python
# WAP to identify the longest word and shortest word in a given sentence
# WAP to identify the most repated word in a given stentence
# WAP out of 3 strings 'hyd','blr','mumbai' which is greater and which is lowest
```

```python
s1='virat.kohli@blr.com,Rohit.sharma@mumbai.com,kl.rahul@lucknow.com'
# {'first_name':['virat','Rohit','KL'],
# 'second_name':['kohli','sharma','rahul'],
# 'company':['blr','mumbai','lucknow']}
d={}
fname=[i[:i.find(".")] for i in s1.split(",")]
mname=[i[i.find(".")+1:i.find("@"):] for i in s1.split(",")]
lname=[i[i.find("@")+1:i.find(".",i.find(".")+1)] for i in s1.split(",")]
d["fname"]=fname
d["mname"]=mname
d["lname"]=lname
d
```

```python
s1='virat.kohli@blr.com,Rohit.sharma@mumbai.com,kl.rahul@lucknow.com'
l1=s1.split(",")
fname=[]
mname=[]
lname=[]
d={}
for i in l1:
    fname.append(i[:i.find(".")])
    mname.append(i[i.find(".")+1:i.find("@")])
    lname.append(i[i.find("@")+1:i.find(".",i.find(".")+1)])
    d["fname"]=fname
    d["mname"]=mname
    d["lname"]=lname
d
```

**word-frequency**

```python
str1='can can you canner can you able to can canner'
#{'can':4,'you':2,'canner':2,'able':1,'to':1}

# d={}
# step-1: split the str1 ======= > you will get a list
# step-2: iterate through loop ==== > each word will print
# step-3: list1.count(<word>): number
# step-4: make the dictionary
d={}
l=str1.split()
for i in l:
    d[i]=l.count(i)
d
```

```python
key=list(d.keys())
val=list(d.values())
key[val.index(max(val))]
```
