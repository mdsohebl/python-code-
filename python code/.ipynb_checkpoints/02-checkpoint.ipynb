**exception handling**

- There are Two block one is "try" another is "exception" block
- your code will be there in "try" block 
- error capture will happens in "exception" block


- try block lets you test a block of code for errors.

- except block lets you handle the error.

- else block lets you execute code when there is no error.

- finally block lets you execute code, regardless of the result of the try- and except blocks

- suppose you have written 500 lines of code at 200 line you got an error this error is migh not
  impact your code so you want execute after 200 line also

- You can define as many exception blocks as you want, e.g. if you want to execute a special block of code for a special kind of error

```python
try:
    a=10
    b=0
    print(a/b)
except:
    print("zero division error, check code")
    
```

**short hand if else**

```python
# one line statment
a=30
b=20
if a > b: print("a is greater than b")
```

```python
# one line if else statment
a = 5
b = 10
print("A") if a > b else print("B")
```

```python
# You can also have multiple else statements on the same line:
# One line if else statement, with 3 conditions
a = 30
b = 30
print("A") if a > b else print("=") if a == b else print("B")
```

**neasted if**

- You can have if statements inside if statements, this is called nested if statements.

```python
x = 40

if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

**pass statment**
- if statements cannot be empty, but if you for some reason have an if statement with no content, put in the pass statement to avoid getting an error.

```python
a = 33
b = 200

if b > a:
  pass
```

```python
# wap ask the user enter a number
# find the number is even or odd number
# eve number: if you divide with 2 the reminder is zero
# odd number
# in python modulus opertor gives remainder

x=eval(input("Enter a number: "))
if x%2==0:
    print(f"{x} is a even number")
else:
    print(f"{x} is a odd number")
```

```python
# using try and exception block
try:
    x=eval(input("Enter a number:"))
    if x%2==0:
        print("{} is a even number".format(x))
    else:
        print("{} is a odd number".format(x))
except Exception as e:
    print(e)
```

```python
# WAP get a random number between 1 to 200
# print you won, if number between 10 to 99: num>=10 and num<99
# otherwise print you lost
import random as r
x=r.randint(1,200)
if x>=10 and x<99:  # same as (10<num<99)
    print(f"you Win because number {x} between 10 to 99")
else:
    print(f"you lost because number {x} not between 10 to 99")
    
```
