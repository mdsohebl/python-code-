**tuple**
- A tuple is a collection which is ordered and unchangeable.and allow dublicates
- they store multiple items in a single variaible
- tuple items are seperated by comma and enclose within ()


```python
# creating tuple
t=(1,2,3,4,3,3)
t
# creating tuple with one items
t1=(1)  # its a int value not a tuple 
print(type(t1))

t2=(1,)
t2
```

- we can create a tuple using tuple method or constructor

```python
t=tuple((1,2,3)) # note a duble bracket
t
```

**access tuple**

```python
t=("hamza","jagga","boss")
t[2]     # Print the second item in the tuple
```

```python
t=("hamza","jagga","boss","lala","soheb")
t[:2]    # its return 1 and 2 value 
```

**change**
- tuple are immutable you cant change tuple value after creation 

```python
t=(1,2,3,2,2)
# t[0]=10    # its not work because it is immutable

# if you want to change tuple value first convert into a list then change 
t=(1,2,3,2,2)
l=list(t)
l[1]=222
t=tuple(l)
t
```

**set**
- A set is a collection which is unordered, unchangeable do not contain dublicate
- set are used to store multople items in a single variable
- Sets are written with curly brackets


```python
s={}     # its not a set it is a dictionary
s1={1,}  # set with one items
s1
```

```python
# creating emty set using set() method or constructor
s=set()
s1=set((1,2,3))
print(type(s))
print(type(s1))
```

```python
# access set item using loop
s={1,2,3,4}
for i in s:
    print(i)
```

**remove-discard**



- remove()
    - Removes the specified item if item does not exist its raise error
- discard()
    -Remove the specified item if item does not exist it will NOT raise an error.

```python
# remove
s={"apple","banana","mango","pomegrant"}
s.remove("banana") # it remove banana
s.remove("ps")     # it will raise error bc pc is not there in set
s
# discard
s={"apple","banana","mango","pomegrant"}
s.discard("banana")  # its also remove banana
s.discard("pen")     # its not raise an error
s


```

**joining two set**
- There are several ways to join two or more sets in Python.

- The union() and update() methods joins all items from both sets.

- The intersection() method keeps ONLY the duplicates.

- The difference() method keeps the items from the first set that are not in the other set(s).

- The symmetric_difference() method keeps all items EXCEPT the duplicates.



**union()-     | is shortcut**
- The union() method returns a new set with all items from both sets

```python
s1={1,2,3}
s2={10,20,30}
s3=s1.union(s2)
s3
```

- use | operator to join two set

```python
s1={1,2,3}
s2={10,20,30}
s3=s1|s2
s3
```

**join multiple set**


```python
s5=s1.union(s2,s3,s4) # first union
s5=s1|s2|s3|s4        # second union

### you can join any data type like list,tuple
```

**update**
- The update() use to a two set changes the original set, and does not return a new set.

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set1.update(set2)
set1
```

**intersection (&) is shortcut**
- The intersection() method will return a new set Join two set, but keep only the duplicates

```python
s1 = {"a", "b" , "c"}
s2 = {"d","b","f","c"}
s3=s1.intersection(s2)
s3

s3=s1&s2   # its a shortcut
s3
```

**intersection_update()**
- The intersection_update() method will update a orignal set Join two set, but keep only the duplicates

```python
s1 = {"a", "b" , "c"}
s2 = {"d","b","f","c"}
s1.intersection_update(s2)
s1&s2
s1
```

**diffrence (-) is shortcut**
- The difference() method will return a new set its Keep all items from set1 that are not in set2

```python
s1 = {"a", "b" , "c"}
s2 = {"d","b","f","c"}
s3=s1.difference(s2)
s1-s2
s3
```

**diffrence-update()**
- The difference_update() method will change orignal set its Keep all items from set1 that are not in set2

```python
s1 = {"a", "b" , "c"}
s2 = {"d","b","f","c"}
s1.difference_update(s2)
s1
```

**symetric_diffrence (^) is a shortcut**
- The symmetric_difference() method will keep only the elements that are NOT present in both sets.
  and return a new set

```python
s1 = {"a", "b" ,"l", "c"}
s2 = {"d","b","f","c"}
s3=s1.symmetric_difference(s2)
s1^s2
s3
```

**symetric_diffrence_update()**
- The symmetric_difference() method will keep only the elements that are NOT present in both sets.but it will change orignal set

```python
s1 = {"a", "b" ,"l", "c"}
s2 = {"d","b","f","c"}
s1.symmetric_difference_update(s2)
s1
```
