**pandas**
- pandas is a python library it is used to analyze a data
- It has functions for analyzing, cleaning, exploring, and manipulating data.

**Why Use Pandas?**

- Pandas allows us to analyze big data and make conclusions based on statistical theories.
- Pandas can clean messy data sets, and make them readable and relevant.


**What Can Pandas Do?**
- Pandas gives you answers about the data. Like:
- Is there a correlation between two or more columns?
- What is average value?
- Max value?
- Min value?

```python
import pandas as pd
d={
    "name":["soheb","lala","hamza"],
    "age":[23,20,40]
}
data=pd.DataFrame(d)
data
```

**series**
- A Pandas Series is like a column in a table.
- It is a one-dimensional array holding data of any type

```python
a=["lala","hamza","jagga"]
data1=pd.Series(a,index=["a","b","c"])   # a,b,c are label can be used to access a specified value.
data1
```

**Dataframe**
- A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.
- Data sets in Pandas are usually multi-dimensional tables, called DataFrames.
- Series is like a column, a DataFrame is the whole table

**loc**
- Pandas use the loc attribute to return one or more specified row(s)

**head()**
- There is also a head() method for viewing the starting rows of the DataFrame.
- The head() method returns the headers and a specified number of rows, starting from the top.


```python
data.head(2) # its give starting two row
```

**tail()**
- There is also a tail() method for viewing the last rows of the DataFrame.
- The head() method returns the headers and a specified number of rows, starting from the top.



```python
data.tail(2)  # its give last two row
```
