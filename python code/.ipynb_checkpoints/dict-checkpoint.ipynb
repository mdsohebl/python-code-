**dic**
- Dictionaries are used to store data values in key:value pairs.
- A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
- that are seperated by commas and enclosed within curly bracket{}

```python
# creating dictionary
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic["name"]             # access value if kay are wrong it gives error
print(dic.get("name1")) # access value using get() method if kay are wrong it gives none but no error
```

**key**
- Returns a list containing the dictionary's keys

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic.keys()
```

**valuse**
- Returns a list of all the values in the dictionary

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic.values()
```

**pop**
- Removes the element with the specified key and return value

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic.pop("age")
```

**popitem**
- Removes the last key-value pair and return pair

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic.popitem()
```

**del**
- The del keyword removes the item with the specified key name and can also delete complete dicnary

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
del dic["age"] # its remove a specific element
del dic        # its remove a complete dicnary
```

**get**
- Returns the value of the specified key

```python
dic={
    "name":"soheb",
    "age":23,
    "add":"Hatai"
}
dic.get("age")
dic["age"]     # its also work
```

**update**
- Updates the dictionary with the specified key-value pairs


```python
d1 = {'a': 1, 'b': 4, 'c': 5}
d2 = {'b': 7, 'd': 9}
d1.update(d2)
d1
```
