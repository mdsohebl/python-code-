**concatnation**
- There are several ways to join, or concatenate, two or more lists in python
    - by using + operator
    - Another way to join two lists is by appending all the items from list2 into list1, one by one:
    - third is using extend method


```python
# # 1 using + operator
# l1=[1,2,3,4]
# l2=["A","B","C","D"]
# l3=l1+l1


# # using appending
# l1=[1,2,3,4]
# l2=["A","B","C","D"]

# for i in l1:
#     l1.append(l2)

# using extend
l1=[1,2,3,4]
l2=["A","B","C","D"]
l1.extend(l2)
l1
```

**list comprension**
- List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list in one line

```python
list1=[1,2,3,4]
o=[i*i for i in list1 ]
o

list2=['hyd','bengaluru','delhi']
o1=[i.capitalize() for i in list2]
o1

list3=['h#d','beng#luru','delhi']
o3=[i for i in list3 if "#" in i]
o3

list4=['abcd','abc','ac','a']
# elemnts less than 3
o4=[i for i in list4 if len(i)<3]
o4
```

**count**
- Returns the number of elements with the specified value

```python
# using count
list1=[1,2,3,'A','B','B','B','C']
list1.count("B")

# using developer mode
list1=[1,2,3,'A','B','B','B','C']
c=0
for i in list1:
    if i=="B":
        c+=1
c
```

- The del is a keyword its removes the specified index or its can also delete the list completely.

**zip**


```python
list1=['Ram','Robert','Raheem']
list2=[25,30,35]

# 'Ram age is 25'
# 'Robert age is 30'
# 'Raheem age is 35'

for i,j in zip(list1,list2):
    print(f"{i} age is {j}")
```

```python
list1=[100,200,300]
list2=[25,30,35]
# output=[125,230,335]

o=[]
for i,j in zip(list1,list2):
    o.append(i+j)
o
```
