**for loop**
- for loop is used for iterating over a sequence of iterable object like      list,tuple,string,dictionary,set
- iterater 
    - An iterator is an object that contains a countable number of values.

```python
- any loop we need three things
    - Intialization  (means= 0)
    - condition
    - incement/decrement
```

```python
# Print each fruit in a fruit list
l=["Apple","banana","mango"]
for i in l:
    print(i)
```

**looping through a string**
- strings are iterable objects, they contain a sequence of characters

```python
# Loop through the letters in the word "soheb"
for i in "soheb":
    print(i)
```

**Range func**
- The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

```python
#  range(6) is not the values of 0 to 6, but the values 0 to 5.
for i in range(6):
    print(i)
```

```python
for i in range(20):
    print(i,end=" ")
```

```python
for i in range(2,20,2): # (start,stop,step)
    print(i)
```

**break statment**
- With the break statement we can stop the loop before it has looped through all the items

```python
l=["apple","google","tcs","infosys"]
for i in l:
    print(i)
    if i=="tcs":  # this time tcs will print
        break    # Exit the loop when x is "tcs"
```

```python
l=["apple","google","tcs","infosys"]
for i in l:
    if i=="tcs":  # this time tcs will not print
        break
    print(i)
```

**continue statment**
- with the continue statement we can stop the current iteration of the loop, and continue with the next:



```python
l=["apple","google","tcs","infosys"]
for i in l:
    if i=="tcs":  # do not print tcs
        continue
    print(i)
```
