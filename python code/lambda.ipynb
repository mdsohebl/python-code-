**Lambda**
- A lambda function is a small anonymous function
- it can take any number of arguments, but can only have one expression it is define using lambda keyword

```python
# one argument
sq=lambda x:x*x
sq(5)

# two argument
sm=lambda x,y:x+y
sm(3,2)

# three argument
men=lambda x,y,z:(x+y+z)/3
men(2,5,8)
```

- Use that function definition to make a function that always doubles the number you send in:


```python
def mul(x):
    return lambda a:a*x
z=mul(2)
print(z(10))
```

```python
# greater number using lambda
g=lambda a,b:(a if a>b else b )
g(30,20)
```

**map()**
- map() function returns a map object(which is an iterator) of the results after applying the given function to each item of a given iterable (list, tuple etc.)

```python
list1=['hyd','mumbai','chennai']
#output: ['Hyd','Mumbai','Chennai']

l=list((map(lambda i:i.capitalize(),list1)))
l
```

```python
list1=[1,2,3,4,5]
# [1,4,9,16,25]
l=list(map(lambda i:i*i,list1))
l
```

**filter**
- The filter() method filters the given sequence with the help of a function that tests each element in the sequence to be true or not. 

```python
list1=['h#d','mum#bai','chennai']
l=list(map(lambda i:"#" in i,list1))
l
```

```python
list1=['h#d','mum#bai','chennai']
l=filter(lambda i: "#" in i ,list1)
list(l)

```
