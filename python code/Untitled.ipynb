```python
class emp:
    company="google"
    no_of_emp=0
    def __init__(self,n):
        self.name=n
        self.raise_am=0.2
        emp.no_of_emp+=1
    def show(self):
        print(f"The emp no {self.no_of_emp} and name is {self.name} raise amount {self.raise_am} company is {self.company}")
        
o1=emp("soheb")
o1.show()
o2=emp("lala")
o2.raise_am=0.3
o2.show()
```

```python
class emp:
    company="google"
    def __init__(self,n):
        self.name=n
    
    @classmethod
    def show(self,cc):
        self.company=cc
c=emp("soheb")
c.company="tesla"
print(c.company)
    
```

```python
class emp:
    def __init__(self,n,a):
        self.name=n
        self.age=a
    @classmethod
    def show(self,s):
        return self(s.split("-")[0],
                   int(s.split("-")[1]))
e=emp.show("soheb-123")
print(e.name)
print(e.age)
```

```python
class emp:
    def __init__(self,n,s):
        self.name=n
        self.salary=s
    @classmethod
    def show(self,st):
        name,salary=st.split(",")
        return self(name,int(salary))
e=emp.show("soheb,2000")
print(e.name)
print(e.salary)
```

```python
# single inheratance 
class parent:
    def __init__(self,n,a):
        self.name=n
        self.age=a
    def show(self):
        print("parent")
class child(parent):
    def show(self):
        print("child")
a1=parent("soheb",22)
a2=child("lala",25)
a1.show()
```

```python
class supers:
    def hello(self):
        print("my name is soheb")
class child(supers):
    def hello1(self):
        print("my name is lala")
        super().hello()
a=child()
a.hello1()
```

```python
class parent:
    def __init__(self,n,a):
        self.name=n
        self.age=a
    def show(self):
        print("This is parent class")
        print(f"name is {self.name} and age is {self.age} language is {self.language}")
class child(parent):
    def show(self):
        print("This is our class")
p=child("soheb",20)
p.show()
```
