**decorator**
- Decorator is a function that takes another func as an argumen and return a new func
  that modify the behavior of the original function the new function is often refered to as a decorator func

```python
def hello(x):
    def fx(*args,**kwargs):
        print("wellcome bro!!!")
        x(*args,**kwargs)
        print("Thanks!!!")
    return fx
@hello
def h1():
    print("soheb")
@hello
def add(a,b):
    print(a+b)
h1()
add(1,2)
```

**getter**
- getter in python are method that are used to access the value of an object properties 
  that are used to return the value of a specefic property 
- typically its define using @ property decorator

```python
class lala:
    def __init__(self,v):
        self._value=v
    def show(self):
        print(f"The value is {self._value}")
    @property
    def ten_v(self):
        return 5*self._value
    @ten_v.setter
    def ten_v1(self,new_v):
        self._value=new_v/2
    
a=lala(10)
print(a.ten_v1)
a.show()
```

**inheritance**
- when a class derive from another class the child class will inherit all the properties and method from parent class

```python
class parent:
    def __init__(self,n,s):
        self.name=n
        self.salary=s
    def show(self):
        print(f"My name is {self.name} and slalary is {self.salary}")
class child(parent):   # inheritance
    def chlang(self):
        print("OK done!!!")
e1=parent("lala",1000)
e1.show()
e2=child("arfat",2000)
e2.show()
```

**operator overloading**


```python
class vector:
    def __init__(self,i,j,k):
        self.i=i
        self.j=j
        self.k=k
    def __str__(self):
        return f"{self.i}i + {self.j}j + {self.k}k"
    def __add__(self,x):
        return vector(self.i+x.i,self.j+x.j,self.k+x.k)
v=vector(2,3,4)
print(v)
v1=vector(1,2,3)
print(v1)
print(v+v1)
```
