```python
# create range fun
class Range:
    def __init__(self,s,e):
        self.start=s
        self.end=e
    def __iter__(self):
        return Range_Iteration(self)

class Range_Iteration:
    def __init__(self,io):
        self.iter_o=io
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.iter_o.start>=self.iter_o.end:
            raise StopIteration
        current=self.iter_o.start
        self.iter_o.start+=1
        return current
o=Range(1,11)
for i in o:
    print(i)
```

```python
#Chaining Generators
def fibb_num(num):
    x,y=0,1
    for _ in range(num):
        x,y=y,x+y
        yield x
    
def square(num):
    for i in num:
        yield i**2

print(sum(square(fibb_num(10))))
```

```python
# Enclosing scope
# Refers to variables in the local scope of an enclosing (outer) function.
def outer():
    a=10    # enclosing var
    def inner():
        print(a)
    inner()
    print("its a outer")
outer()
print("done")
```

```python
# nonlocal scope
def outer():
    a=111
    def inner():
        nonlocal a
        a+=2     # Modify enclosing scope variable
        print("inner var",a)
    inner()
    print("outer var",a)
outer()
```

```python
class Atm:
    def __init__(self):
        self.balance=0
        self.pin=""
        self.menu()
    
    def menu(self):
        user_input=input("""
        Hai how can i help you
        1. Enter for creater pin
        2. Enter for change pin
        3. Enter for check Balance
        4. Enter for cash withraw
        5. Exit
        """)
    
        if user_input=="1":
            self.create_pin()
        elif user_input=="2":
            self.change_pin()
        elif user_input=="3":
            self.check_balance()
        elif user_input=="4":
            self.withraw()
        else:
            exit()
    
    def create_pin(self):
        user_balance=int(input("Enter a balance"))
        self.balance=user_balance
        user_pin=input("Enter a pin")
        self.pin=user_pin
        
        print("pin created succefully")
        self.menu()
    
    def change_pin(self):
        old_pin=input("Enter a old pin")
        
        if self.pin==old_pin:
            new_pin=input("enter a new pin")
            self.pin=new_pin
            print("new pin created successfully")
            self.menu()
            
        else:
            print("Enter valid pin")
            old_pin=input("Enter a old pin")
            print("Enter a valif pin")
            self.menu()
            
    def check_balance(self):
        enter_pin=input("enter a pin")
        if enter_pin==self.pin:
            print(self.balance)
        else:
            print("invalid pin")
    
    def withraw(self):
        enter_pin=input("enter a pin")
        if enter_pin==self.pin:
            enter_amount=int(input("enter a amount"))
            if enter_amount<self.balance:
                print("reaming balance is",self.balance-enter_amount)
            else:
                print("inefficient balalnce")
        else:
            print("invalid pin")
            
        self.menu()
        
        
        
    
a=Atm()
```

```python
# fraction class
class Fraction:
    def __init__(self,n,d):
        self.num=n
        self.den=d
    def __str__(self):
        return f"{self.num}/{self.den}"
    
    def __add__(self,other):
        num=self.num*other.den+ other.num*self.den
        den=self.den*other.den
        return f"{num}/{den}  is {num/den}"
    
    def __sub__(self,other):
        num=self.num*other.den - self.num*other.den
        den=self.den *other.den
        return f"{num}/{den}  is {num/den}"
    
    def __mul__(self,other):
        num=self.num*other.num
        den=self.den*other.den
        return f"{num}/{den} is {num/den}"
    
    def __truediv__(self,other):
        num=self.num*other.den
        den=self.den*other.num
        return f"{num}/{den} is {num/den}"
        
f1=Fraction(2,3)
f2=Fraction(5,6)

print(f1+f2)
print(f1-f2)
print(f1*f2)
print(f1/f2)
```

```python
# 1 A user can create and view 2D coordinates
class Point:
    def __init__(self,x,y):
        self.xcor=x
        self.ycor=y
    def __str__(self):
        return f"<{self.xcor},{self.ycor}>"   # 1 ans
    
    def euclidian_dis(self,other):
        return ((self.xcor-other.xcor)**2+(self.ycor-other.ycor)**2)**0.5  # 2 ans
    
    def distance_origin(self):
        return (self.xcor**2 +self.ycor**2)**0.5    # 3 ans
    
c1=Point(2,6)
c2=Point(2,1)
# c1.euclidian_dis(c2)
# c1.distance_origin()

class Line:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
        
    def __str__(self):
        return f"{self.a}x +{self.b}y +{self.c}z"
    
    def point_liy(self,other):
        if self.a*other.xcor + self.b*other.ycor +self.c==0:
            return "line liye between two line"
        else:
            return "line not liye between two line"
        
    def short_dis(self,other):
        return abs(self.a*other.xcor+self.b*other.ycor*self.c)/(self.a**2 +self.b**2)**0.5
    
l=Line(2,3,3)
# l.point_liy(c2)
l.short_dis(c2)
```

```python
# Exception raise
class Atm:
    def __init__(self,b):
        self.balance=b
    
    def withraw(self,a):
        if a<0:
            raise Exception("balance cant be negative")
        if a>self.balance:
            raise Exception("inefficient balance")
        self.balance=self.balance-a
o=Atm(2000)
try:
    o.withraw(-2)
except Exception as e:
    print(e)
else:
    print(o.balance)
```

```python
class Security(Exception):
    def __init__(self,m):
        print(m)
    def logout(self):
        print("logout")
class Google:
    def __init__(self,n,e,p):
        self.name=n
        self.email=e
        self.password=p
    def login(self,e,p):
        if self.email==e and self.password==p:
            print("login succfully")
        else:
            raise Security("wrong email or password")
g=Google("soheb","soheb@1",123)
try:
    g.login("soheb@",123)
except Security as e:
    e.logout()
else:
    print(g.name)
finally:
    print("close")
```
